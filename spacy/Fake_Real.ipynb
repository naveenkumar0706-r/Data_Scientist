```python
import pandas as pd

df = pd.read_csv('Fake_Real_Data.csv')

print(df.shape)
print(df.head())
df['label'].value_counts()
```

```python
df['label_num'] = df['label'].map({'Fake': 0, 'Real': 1})
print(df.head())
```

```python
import spacy
nlp = spacy.load('en_core_web_lg')
```

```python
df['vector'] = df['Text'].apply(lambda text: nlp(text).vector)
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(df.vector.values, df.label_num, test_size=0.2, random_state=42)
```

```python
import numpy as np

x_train_2d = np.stack(x_train)
x_test_2d = np.stack(x_test)
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import classification_report


scaler = MinMaxScaler()
scaled_train = scaler.fit_transform(x_train_2d)
scaled_test = scaler.transform(x_test_2d)

clf = MultinomialNB()
clf.fit(scaled_train, y_train)
```

```python
y_pred = clf.predict(scaled_test)
print(classification_report(y_test, y_pred))
```

```python
from sklearn.metrics import accuracy_score, classification_report
print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(x_train_2d, y_train)
```

```python
y_pred_knn = knn.predict(x_test_2d)
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred_knn))
print(classification_report(y_test, y_pred_knn))
```

```python
print("Naive Bayes Accuracy:", accuracy_score(y_test, y_pred))
print("KNN Accuracy:", accuracy_score(y_test, y_pred_knn))
if accuracy_score(y_test, y_pred) > accuracy_score(y_test, y_pred_knn):
    print("Best Model: Naive Bayes")
else:
    print("Best Model: KNN")
```
